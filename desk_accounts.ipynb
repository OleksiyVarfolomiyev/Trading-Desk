# Desk accounts

Every account in `accounts.json` as one desk: the combined equivalent of `hl_accounts.ipynb` and `ib_account.ipynb`. HL wallets use their saved value and PnL history; IB uses IB Flex daily NAV with deposits and withdrawals (see `desk/venues/ib_flex.py`). Together they form one desk time-weighted return, so deposits, withdrawals and transfers between accounts are never performance. Wallet tokens on other chains have no value history, so they appear in current totals only. Read-only: nothing here places orders.

TWS should be running for live IB positions; without it the IB history and saved values are still used.
